## 4. Relationships and Final decisions

### Chi-Square Test

```python
table = pd.crosstab(df['department'],df['attrition'])
print(table)

chi2,p,dof,excepted = st.chi2_contingency(table)

print(f'Chi2: {chi2}')
print(f'P-value: {p}')

if p < 0.05:
    print("Department and Attrition are related (p < 0.05).")
else:
    print('There is no strong relation between Department and Atrrition')
```

### Correlation Analysis

```python
corr = df.corr(numeric_only=True, method='spearman')

attrition_corr = corr['attrition_flag'].sort_values(ascending=False)
print(attrition_corr)


fig_corr = go.FigureWidget()

fig_corr.add_trace(
    go.Heatmap(
        z=corr.values,
        x=corr.columns,
        y=corr.index,
        colorscale='RdBu_r',      
        zmin=-1,
        zmax=1,
        text=corr.values,
        texttemplate='%{text:.2f}', 
        textfont=dict(size=9, color='black'),
        hovertemplate='x: %{x}<br>y: %{y}<br>Correlation: %{z:.2f}<extra></extra>',
        colorbar=dict(title='Correlation')
    )
)

fig_corr.add_trace(
    go.Scatter(
        x=[],
        y=[],
        mode='markers',
        marker=dict(
            symbol='square',
            size=28,
            color='rgba(255,255,255,0)',
            line=dict(color='rgba(0,255,0,1)', width=2)  
        ),
        hoverinfo='skip',
        showlegend=False
    )
)

fig_corr.update_layout(
    template='plotly_white',
    title='Correlation Matrix',
    xaxis_title='',
    yaxis_title='',
    height=800,
    width=900,
    hovermode='closest',
    yaxis=dict(autorange='reversed')  
)

def highlight_hover(trace, points, selector):
    if len(points.xs) == 0:
        return
    x_val = points.xs[0]
    y_val = points.ys[0]
    fig_corr.data[1].x = [x_val]
    fig_corr.data[1].y = [y_val]

fig_corr.data[0].on_hover(highlight_hover)
fig_corr.show()

```

###  Final decisions

#### Answer

**5 key findings:**

1. The total Attrition Rate was 19.97% (1,997 out of 10,000 employees left the company).
2. Employees who work overtime have a slightly higher attrition rate (20.07% vs 19.87%), but the difference is small and should not be treated as a strong explanation on its own.
3. Job Satisfaction and Work Life Balance scores were actually slightly `HIGHER` among employees who left, and only Work Environment Satisfaction was lower so this shows no clear, consistent trend was found across these three metrics.
4. There are small differences in attrition rate by Department (Finance 20.85%) and Job Role (Assistant 21.43%), but the Chi-Square test showed this is not statistically significant.
5. Chi-Square test result: p=0.748 (p ≥ 0.05) — no statistically significant relationship was found between Department and Attrition.

**3 business recommendations:**

1. We need to drop overtime load. To do this we must do workload disturbition across all teams.
2. Conduct quarterly employee satisfaction surveys, identify the main concerns within low-satisfaction teams, and implement targeted improvement actions based on the findings.
3. Group the main turnover drivers into key categories such as compensation, workload, career development, job satisfaction, management, and work environment, then develop targeted retention actions for the highest-impact factors.
